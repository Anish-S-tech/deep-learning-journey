```python
# This file explains about the implementation of forward propogation which is an essential process in Neural Networks
```

```python
# Step 1: Importing necessary libraries

import numpy as np
import pandas as pd
```

```python
# Step 2: Create a sample dataset

data = {
    'CGPA': [8.4,9.1,7.8],
    'profile_score': [85,92,78],
    'lpa': [12,10,8]
}

df = pd.DataFrame(data)
X = df[['CGPA', 'profile_score']].values
```

```python
# Step 3: Initializing the parameters

def initialize_parameters():
  np.random.seed(1)
  W = np.random.randn(2,1)*0.01
  b = np.zeros((1,1))
  return W,b
```

```python
# Step 4: Define forward propogation

def forward_propogation(X,W,b):
  Z = np.dot(X,W)+b
  A = 1/(1+np.exp(-Z))
  return A
```

```python
# Step 5: Printing the final result of forward propogation

W, b = initialize_parameters()
A = forward_propogation(X,W,b)
print("Final output is\n",A)
```
